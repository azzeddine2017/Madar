## 🪄 Step 3: Register `%%madar` Interactive Cell Magic
Register the `%%madar` (and `%%ring`) IPython cell magic so you can write and run native Madar code directly inside Jupyter notebook cells without shell overhead.

```python
from IPython.core.magic import register_cell_magic
import subprocess, tempfile, os, sys

MADAR_BIN = os.path.abspath("bin/ring")

@register_cell_magic
def madar(line, cell):
    """IPython Cell Magic to execute Madar code and display outputs directly."""
    with tempfile.NamedTemporaryFile("w", suffix=".ring", delete=False, encoding="utf-8") as f:
        f.write(cell)
        tmp_file = f.name
    
    try:
        env = os.environ.copy()
        env["LD_LIBRARY_PATH"] = os.path.abspath("lib") + ":" + env.get("LD_LIBRARY_PATH", "")
        cwd = os.path.abspath("/content/Madar") if os.path.exists("/content/Madar") else os.path.abspath(".")
        res = subprocess.run([MADAR_BIN, tmp_file], capture_output=True, text=True, env=env, cwd=cwd)
        if res.stdout:
            print(res.stdout, end="")
        if res.stderr:
            print(res.stderr, end="", file=sys.stderr)
    finally:
        if os.path.exists(tmp_file):
            os.remove(tmp_file)

ip = get_ipython()
ip.register_magic_function(madar, 'cell', 'ring')
print("✅ %%madar and %%ring cell magics registered successfully!")
```

## ✒️ Step 6: Hybrid Knowledge Retrieval with AlQalam (C++20 Hybrid RAG)
The **AlQalam** extension blends dense vector similarity search (70%) with inverted lexical indexing `QalamIndex` (30%) alongside ultra-fast binary serialization.

```python
%%madar
load "stdlibcore.ring"
load "libraries/madar_ai/madar_ai.ring"
load "libraries/madar_ai/rag.ring"
load "alqalam.ring"

cModelPath = "models/qwen2.5-7b-instruct-q4_0.gguf"
if not fexists(cModelPath) cModelPath = "models/deepseek-r1-distill-qwen-7b-q4_0.gguf" ok

oTimer = new QalamChronos()
oTimer.reset()

see "[RAG] Initializing language model for RAG... " + nl
llm = ai_open(cModelPath)
llm.setGPU(1)
llm.setThreads(4)
store = new VectorStore(llm)

# Index sample corpus using AlQalam
store.addText("Madar runtime supports GPU acceleration via OpenCL zero-copy and AVX2 SIMD vectorization.")
store.addText("AlQalam C++20 library provides high-performance data structures including QalamIndex and dense vectors.")
store.addText("The capital of France is Paris, famous for the Eiffel Tower and rich cultural heritage.")

cQuestion = "What are the key capabilities of the AlQalam library in Madar?"
see nl + "Question: " + cQuestion + nl

# Execute Hybrid Search (70% semantic + 30% lexical via AlQalam)
oTimer.reset()
aResults = store.hybridSearch(cQuestion, 2, 0.7)
see f"[RAG] Hybrid Search Results (Time: {oTimer.elapsed()}):" + nl

for res in aResults
    see f"  - [Score: {res[1]}] (Semantic: {res[3]} | Lexical: {res[4]}) {res[2].cText}" + nl
next

best_text = aResults[1][2].cText
see nl + "[RAG] Generating grounded answer based on retrieved context:" + nl
cPrompt = f"Answer concisely based strictly on this context: {best_text}\nQuestion: {cQuestion}"
see llm.generate(cPrompt, 256) + nl
```

## 💬 Step 8: Interactive In-Notebook Chat Widget
Engage in interactive conversation with the mounted AI model inside Colab using a styled UI widget.

```python
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output
import subprocess, os, tempfile

chat_history = []

out_area = widgets.Output()
text_input = widgets.Text(
    placeholder='Type your question to Madar AI here and press Enter...',
    layout=widgets.Layout(width='75%')
)
send_btn = widgets.Button(
    description='Send 🚀',
    button_style='primary',
    layout=widgets.Layout(width='20%')
)

def ask_madar(prompt):
    ring_code = f'''
load "libraries/madar_ai/madar_ai.ring"
cModel = "models/qwen2.5-7b-instruct-q4_0.gguf"
if not fexists(cModel) cModel = "models/deepseek-r1-distill-qwen-7b-q4_0.gguf" ok
model = ai_open(cModel)
model.setGPU(1)
model.setThreads(4)
model.setTemperature(0.7)
model.setMaxTokens(200)
see model.generate("{prompt}", 200)
'''
    with tempfile.NamedTemporaryFile("w", suffix=".ring", delete=False, encoding="utf-8") as f:
        f.write(ring_code)
        tmp_name = f.name
    
    try:
        env = os.environ.copy()
        env["LD_LIBRARY_PATH"] = os.path.abspath("lib") + ":" + env.get("LD_LIBRARY_PATH", "")
        cwd = os.path.abspath("/content/Madar") if os.path.exists("/content/Madar") else os.path.abspath(".")
        res = subprocess.run([os.path.abspath("bin/ring"), tmp_name], capture_output=True, text=True, env=env, cwd=cwd)
        return res.stdout.strip()
    finally:
        if os.path.exists(tmp_name):
            os.remove(tmp_name)
            
def on_send(b):
    user_msg = text_input.value.strip()
    if not user_msg:
        return
    text_input.value = ''
    
    with out_area:
        display(HTML(f'<div style="background-color:#e3f2fd; padding:10px; border-radius:8px; margin:5px 0;"><b>🧑 You:</b> {user_msg}</div>'))
        display(HTML('<div id="loading" style="color:#888;"><i>⏳ Madar is thinking...</i></div>'))
    
    bot_msg = ask_madar(user_msg)
    
    with out_area:
        clear_output(wait=True)
        for role, text in chat_history:
            bg = '#e3f2fd' if role == 'user' else '#f1f8e9'
            author = '🧑 You' if role == 'user' else '🪐 Madar'
            display(HTML(f'<div style="background-color:{bg}; padding:10px; border-radius:8px; margin:5px 0;"><b>{author}:</b> {text}</div>'))
        
        display(HTML(f'<div style="background-color:#f1f8e9; padding:10px; border-radius:8px; margin:5px 0;"><b>🪐 Madar:</b> {bot_msg}</div>'))
    
    chat_history.append(('user', user_msg))
    chat_history.append(('bot', bot_msg))

send_btn.on_click(on_send)
text_input.on_submit(on_send)

controls = widgets.HBox([text_input, send_btn])
display(widgets.VBox([out_area, controls]))
```
